# Overview
Middleware lets you tightly control what happens inside an agent's run loop — reshaping prompts, retrying failures, enforcing limits, and adding guardrails without rewriting the agent.

## Concept
An agent loops between calling the model and running tools. Middleware exposes hooks before and after each of those steps, plus wrapping hooks that fully surround a model or tool call. You attach middleware as a list to create_agent; they compose in order, so the first listed wraps the outermost layer.

- Node hooks: before_agent, before_model, after_model, after_agent
- Wrap hooks: wrap_model_call, wrap_tool_call
